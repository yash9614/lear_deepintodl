# Training Algorithms — Bonus: Manual Optimizer Implementations Clean Solution

From-scratch implementations of Gradient Descent, Momentum, RMSProp, and Adam.

**Key concepts:** Update rules for GD, Momentum, RMSProp, Adam (bias correction).

## Problem 1: Basic Gradient Descent\n$L(w) = (w-4)^2$

```python
learning_rate = 0.1
w = 0.0
for i in range(10):
    gradient = 2 * (w - 4)
    w = w - learning_rate * gradient
    print(f"Step {i+1}: w = {w:.4f}")
```

## Problem 2: Momentum\n$f(x,y) = x^2 + 3y^2$

```python
def gradient(x, y):
    return 2*x, 6*y

x, y = 2.0, 2.0
learning_rate = 0.1
momentum = 0.9
vx, vy = 0.0, 0.0

for i in range(5):
    dx, dy = gradient(x, y)
    vx = momentum * vx - learning_rate * dx
    vy = momentum * vy - learning_rate * dy
    x += vx
    y += vy
    print(f"Step {i+1}: x = {x:.4f}, y = {y:.4f}")
```

## Problem 3: RMSProp\n$f(w) = w^2 + 5$

```python
def gradient(w):
    return 2 * w

w = 5.0
learning_rate = 0.01
beta = 0.9
epsilon = 1e-8
squared_gradient_average = 0.0

for i in range(15):
    grad = gradient(w)
    squared_gradient_average = beta * squared_gradient_average + (1 - beta) * (grad ** 2)
    w = w - learning_rate * grad / (squared_gradient_average ** 0.5 + epsilon)
    print(f"Step {i+1}: w = {w:.4f}")
```

## Problem 4: Adam\n$f(x) = x^4 - 3x^3 + 2$

```python
def gradient(x):
    return 4*x**3 - 9*x**2

x = 3.0
learning_rate = 0.01
beta1, beta2 = 0.9, 0.999
epsilon = 1e-8
m, v = 0.0, 0.0

for t in range(1, 20):
    grad = gradient(x)
    m = beta1 * m + (1 - beta1) * grad
    v = beta2 * v + (1 - beta2) * (grad ** 2)
    m_hat = m / (1 - beta1 ** t)
    v_hat = v / (1 - beta2 ** t)
    x = x - learning_rate * m_hat / (v_hat ** 0.5 + epsilon)
    print(f"Step {t}: x = {x:.4f}")
```
