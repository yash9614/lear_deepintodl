## Task 1 — Inventory management
3 days × 5 items. Rows are days, columns are items.

```python
inventory = torch.tensor(
    [[12, 15, 10, 0, 5],
     [10,  8,  7, 5, 4],
     [20, 10, 15, 5, 2]],
    dtype=torch.float32,
)
print(inventory)
print('shape', tuple(inventory.shape))
```

```python
# total of each item across all days → sum over rows (dim 0)
item_totals = inventory.sum(dim=0)
print('total per item:', item_totals)

# average inventory per day → mean over items (dim 1)
avg_per_day = inventory.mean(dim=1)
print('avg per day:', avg_per_day)
```

## Task 2 — Vehicle flow
Two roads, 24 hours. Simulate with random integers.

```python
torch.manual_seed(0)
road_a = torch.randint(10, 80, (1, 24)).float()
road_b = torch.randint(10, 80, (1, 24)).float()
print('road A shape', tuple(road_a.shape), 'road B shape', tuple(road_b.shape))
print('A first 6 hours', road_a[0, :6])
print('B first 6 hours', road_b[0, :6])
```

```python
def total_per_road(a, b):
    return torch.stack([a.sum(), b.sum()])

def total_per_hour(a, b):
    return (a + b).squeeze(0)

print('day totals [A, B]:', total_per_road(road_a, road_b))
print('hourly totals:', total_per_hour(road_a, road_b))
```

## Task 3 — Fitness matrix
Rows = members. Columns = strength, stamina, flexibility.

```python
fitness = torch.tensor(
    [[7.0, 6.0, 5.0],
     [8.0, 5.5, 6.5],
     [6.0, 7.5, 8.0]]
)
weights = torch.tensor([0.8, 1.2, 1.5])  # broadcast across columns
weighted = fitness * weights
print('weighted scores\n', weighted)

print('row max (best metric per member):', weighted.max(dim=1).values)
print('col max (best member per metric):', weighted.max(dim=0).values)

t = weighted.T
print('transposed shape', tuple(t.shape), '→ rows are now metrics, cols are members')
```

## Task 4 — Chain rule / autograd
$y = 2x^3 + 5x^2 - 3x + 7$ at $x = 4$

Analytic gradient: $dy/dx = 6x^2 + 10x - 3$

```python
x = torch.tensor(4.0, requires_grad=True)
y = 2 * x**3 + 5 * x**2 - 3 * x + 7
print('y =', y.item())          # 2*64 + 5*16 - 12 + 7 = 203

y.backward()
print('dy/dx =', x.grad.item())  # 6*16 + 10*4 - 3 = 133
```

## Task 5 — Camera calibration

```python
camera1 = torch.tensor([[1., 2.], [3., 4.]])
camera2 = torch.tensor([[5., 6.], [7., 8.]])

print('Hadamard (element-wise)\n', camera1 * camera2)
print('matmul (dot / matrix product)\n', camera1 @ camera2)
print('det cam1 =', torch.det(camera1).item())  # 1*4 - 2*3 = -2
print('det cam2 =', torch.det(camera2).item())  # 5*8 - 6*7 = -2
```

## Task 6 — Tiny perceptron  $y = wx + b$

```python
w = torch.tensor(2.0)
b = torch.tensor(1.0)
x = torch.tensor([1., 2., 3., 4.])
y = w * x + b
print(y)  # [3, 5, 7, 9]
```
