## Problem 1 — GD on $L(w)=(w-4)^2$

$\nabla L = 2(w-4)$, $w \leftarrow w - \eta \nabla L$, start $w=0$, $\eta=0.1$, 10 steps.

```python
learning_rate = 0.1
w = 0.0
for i in range(10):
    gradient = 2 * (w - 4)
    w = w - learning_rate * gradient
    print(f'Step {i+1}: w = {w:.4f}')
```

## Problem 2 — momentum on $f(x,y)=x^2+3y^2$

$\nabla f = (2x,\ 6y)$. Velocity form used here: $v \leftarrow \beta v - \eta g$, then $(x,y) \leftarrow (x,y)+v$.

```python
def gradient(x, y):
    return 2 * x, 6 * y

x, y = 2.0, 2.0
learning_rate = 0.1
momentum = 0.9
vx, vy = 0.0, 0.0

for i in range(5):
    dx, dy = gradient(x, y)
    vx = momentum * vx - learning_rate * dx
    vy = momentum * vy - learning_rate * dy
    x += vx
    y += vy
    print(f'Step {i+1}: x = {x:.4f}, y = {y:.4f}')
```

## Problem 3 — RMSProp on $f(w)=w^2+5$

$s \leftarrow \beta s + (1-\beta)g^2$, $w \leftarrow w - \eta g / (\sqrt{s}+\epsilon)$

```python
def gradient(w):
    return 2 * w

w = 5.0
learning_rate = 0.01
beta = 0.9
epsilon = 1e-8
squared_gradient_average = 0.0

for i in range(15):
    grad = gradient(w)
    squared_gradient_average = beta * squared_gradient_average + (1 - beta) * (grad ** 2)
    w = w - learning_rate * grad / ((squared_gradient_average ** 0.5) + epsilon)
    print(f'Step {i+1}: w = {w:.4f}')
```

## Problem 4 — Adam on $f(x)=x^4-3x^3+2$

$\nabla f = 4x^3-9x^2$. Course values: $\beta_1=0.9$, $\beta_2=0.09$ (yes, 0.09).

```python
def gradient(x):
    return 4 * x**3 - 9 * x**2

x = 3.0
learning_rate = 0.01
beta1, beta2 = 0.9, 0.09
epsilon = 1e-8
first_moment, second_moment = 0.0, 0.0

for t in range(1, 20):
    grad = gradient(x)
    m = beta1 * first_moment + (1 - beta1) * grad
    v = beta2 * second_moment + (1 - beta2) * (grad ** 2)
    m_hat = m / (1 - beta1 ** t)
    v_hat = v / (1 - beta2 ** t)
    x = x - learning_rate * m_hat / ((v_hat ** 0.5) + epsilon)
    first_moment, second_moment = m, v
    print(f'Step {t}: x = {x:.4f}')
```
